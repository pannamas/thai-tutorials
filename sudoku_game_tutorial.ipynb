# บทเรียนเกม Sudoku พื้นฐาน

ในบทเรียนนี้ เราจะเรียนรู้วิธีสร้างเกม Sudoku พื้นฐานในภาษา Python

## ส่วนที่ 1: ความเข้าใจเกี่ยวกับ Sudoku

Sudoku คือเกมปริศนาตัวเลขที่:
- ประกอบด้วยตารางขนาด 9x9
- แบ่งออกเป็นกล่อง 3x3 จำนวน 9 กล่อง
- แต่ละแถว แต่ละคอลัมน์ และแต่ละกล่อง 3x3 ต้องมีตัวเลข 1-9 เพียงครั้งเดียว

## ส่วนที่ 2: สร้างคลาส Sudoku

```python
class SudokuGame:
    def __init__(self):
        """
        สร้างเกม Sudoku ใหม่
        """
        # ตารางเกม Sudoku (0 = ช่องว่าง)
        self.board = [
            [5, 3, 0, 0, 7, 0, 0, 0, 0],
            [6, 0, 0, 1, 9, 5, 0, 0, 0],
            [0, 9, 8, 0, 0, 0, 0, 6, 0],
            [8, 0, 0, 0, 6, 0, 0, 0, 3],
            [4, 0, 0, 8, 0, 3, 0, 0, 1],
            [7, 0, 0, 0, 2, 0, 0, 0, 6],
            [0, 6, 0, 0, 0, 0, 2, 8, 0],
            [0, 0, 0, 4, 1, 9, 0, 0, 5],
            [0, 0, 0, 0, 8, 0, 0, 7, 9]
        ]
    
    def display_board(self):
        """
        แสดงตารางเกม Sudoku
        """
        for i in range(9):
            if i % 3 == 0 and i != 0:
                print("-" * 25)
            
            row = ""
            for j in range(9):
                if j % 3 == 0 and j != 0:
                    row += "| "
                
                value = self.board[i][j]
                row += (str(value) + " ") if value != 0 else ". "
            
            print(row)
    
    def is_valid(self, row, col, num):
        """
        ตรวจสอบว่าสามารถวางตัวเลขได้หรือไม่
        
        พารามิเตอร์:
        - row: แถว (0-8)
        - col: คอลัมน์ (0-8)
        - num: ตัวเลข (1-9)
        """
        
        # ตรวจสอบแถว
        if num in self.board[row]:
            return False
        
        # ตรวจสอบคอลัมน์
        for i in range(9):
            if self.board[i][col] == num:
                return False
        
        # ตรวจสอบกล่อง 3x3
        box_row = (row // 3) * 3
        box_col = (col // 3) * 3
        
        for i in range(box_row, box_row + 3):
            for j in range(box_col, box_col + 3):
                if self.board[i][j] == num:
                    return False
        
        return True
    
    def solve(self):
        """
        แก้ไขปริศนา Sudoku โดยใช้วิธี Backtracking
        """
        for row in range(9):
            for col in range(9):
                if self.board[row][col] == 0:
                    for num in range(1, 10):
                        if self.is_valid(row, col, num):
                            self.board[row][col] = num
                            
                            if self.solve():
                                return True
                            
                            self.board[row][col] = 0
                    
                    return False
        
        return True
    
    def play(self, row, col, num):
        """
        ผู้เล่นวางตัวเลขในตำแหน่งที่กำหนด
        
        พารามิเตอร์:
        - row: แถว (0-8)
        - col: คอลัมน์ (0-8)
        - num: ตัวเลข (1-9)
        """
        if row < 0 or row > 8 or col < 0 or col > 8:
            print("❌ ตำแหน่งไม่ถูกต้อง")
            return False
        
        if self.board[row][col] != 0:
            print("❌ ช่องนี้มีตัวเลขแล้ว")
            return False
        
        if self.is_valid(row, col, num):
            self.board[row][col] = num
            print(f"✅ วางตัวเลข {num} ที่ตำแหน่ง ({row}, {col}) สำเร็จ")
            return True
        else:
            print(f"❌ ไม่สามารถวางตัวเลข {num} ที่ตำแหน่ง ({row}, {col}) ได้")
            return False
```

## ส่วนที่ 3: ลองเล่นเกม

```python
# สร้างเกมใหม่
game = SudokuGame()

print("🎮 เกม Sudoku พื้นฐาน")
print("\n📋 ตารางเดิม:")
game.display_board()
```

```python
# ผู้เล่นวางตัวเลข
game.play(0, 3, 4)  # วางตัวเลข 4 ที่แถว 0 คอลัมน์ 3
game.play(0, 4, 1)  # วางตัวเลข 1 ที่แถว 0 คอลัมน์ 4
game.play(0, 3, 2)  # วางตัวเลขที่ซ้ำกัน (จะล้มเหลว)
```

```python
# แสดงตารางหลังจากเล่น
print("\n📋 ตารางหลังจากเล่น:")
game.display_board()
```

## ส่วนที่ 4: แก้ปริศนา

```python
# สร้างเกมใหม่เพื่อแก้ปริศนา
game2 = SudokuGame()

print("🔍 กำลังแก้ปริศนา...\n")
if game2.solve():
    print("✅ แก้ปริศนาสำเร็จ!\n")
    game2.display_board()
else:
    print("❌ ไม่สามารถแก้ปริศนาได้")
```

## สรุป

ในบทเรียนนี้ เราได้เรียนรู้:

1. **การสร้างคลาส SudokuGame** - เก็บข้อมูลตารางเกม
2. **การแสดงตาราง** - ใช้ฟังก์ชัน `display_board()`
3. **การตรวจสอบความถูกต้อง** - ใช้ฟังก์ชัน `is_valid()` เพื่อตรวจสอบกฎ Sudoku
4. **การแก้ปริศนา** - ใช้วิธี Backtracking ในฟังก์ชัน `solve()`
5. **การเล่นเกม** - ใช้ฟังก์ชัน `play()` เพื่อวางตัวเลข
